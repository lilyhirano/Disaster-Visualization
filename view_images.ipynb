```python
import json
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
from starter_code.data_utils import get_images, get_labels
```

```python
disaster_list = ["hurricane-matthew", "midwest-flooding", "socal-fire"]
```

```python
data = {}
split = "train"
with open('config.json') as config_file:
    config = json.load(config_file)
    data_dir = config['data_dir']

for disaster in disaster_list:
    print(f"Loading {split} images and labels for {disaster} dataset...")
    images = get_images(data_dir, disaster, split=split)
    labels = get_labels(data_dir, disaster, split=split)
    data[disaster] = {"images": images, "labels": labels}
```

```python
disaster = disaster_list[0]
images = data[disaster]["images"]
labels = data[disaster]["labels"]


fig, axs = plt.subplots(4, 5, figsize=(20, 10))
axs = axs.flatten()

#load 20 imgs
for idx in range(20):
    img = images[idx]
    label = labels[idx]

    axs[idx].set_title(f"{disaster}, image {idx}, label {label}")
    axs[idx].imshow(img.astype(np.uint8))
    axs[idx].axis("off")


plt.show()
```
